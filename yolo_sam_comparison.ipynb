```python
import cv2
import numpy as np
import torch
from google.colab.patches import cv2_imshow  #  Import for Colab
from ultralytics import YOLO
from segment_anything import SamPredictor, sam_model_registry

#  Load YOLO Model
yolo_model = YOLO("/content/content/runs/detect/train5/weights/best.torchscript")

#  Load SAM Model
sam_checkpoint = "/content/sam_vit_h_4b8939.pth"  # Update the path if needed
sam = sam_model_registry["vit_h"](checkpoint=sam_checkpoint)
sam_predictor = SamPredictor(sam)

def yolo_sam_hybrid(image_path):
    # Load image
    image = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Run YOLO detection
    results = yolo_model.predict(image_rgb, conf=0.5)

    # Extract YOLO detections (bounding boxes)
    for result in results:
        boxes = result.boxes.xyxy.cpu().numpy()  # Convert to NumPy array
        for box in boxes:
            x1, y1, x2, y2 = map(int, box)  # Convert coordinates

            # Set SAM input image
            sam_predictor.set_image(image_rgb)

            # Create mask with SAM using bounding box from YOLO
            input_box = np.array([x1, y1, x2, y2])
            masks, _, _ = sam_predictor.predict(box=input_box[None, :], multimask_output=False)

            # Draw YOLO bounding box
            cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Apply segmentation mask
            mask = masks[0]
            image[mask > 0] = [0, 0, 255]  # Red overlay for segmented area

    return image

#  Test on an image
result_img = yolo_sam_hybrid("/content/runs/detect/predict/frame_0115.jpg")

#  Save result
cv2.imwrite("yolo_sam_output.jpg", result_img)

#  Display in Colab
cv2_imshow(result_img)  # No window name needed in Colab
```

```python
import cv2
import matplotlib.pyplot as plt

# Test image path
image_path = "/content/drive/MyDrive/merged_dataset/images/test/frame_0070.jpg"  # Change filename

# Get YOLO output
yolo_result = yolo_model.predict(image_path, conf=0.5)[0].plot()

# Get YOLO-SAM Hybrid output
yolo_sam_result = yolo_sam_hybrid(image_path)

# Show both outputs side by side
fig, axs = plt.subplots(1, 2, figsize=(12, 6))

axs[0].imshow(yolo_result)
axs[0].set_title("YOLO Detection")

axs[1].imshow(yolo_sam_result)
axs[1].set_title("YOLO-SAM Hybrid")

plt.show()
```
