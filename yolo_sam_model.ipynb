```python
import cv2
import numpy as np
import torch
from google.colab.patches import cv2_imshow  #  Import for Colab
from ultralytics import YOLO
from segment_anything import SamPredictor, sam_model_registry

#  Load YOLO Model
yolo_model = YOLO("/content/content/runs/detect/train5/weights/best.torchscript")

#  Load SAM Model
sam_checkpoint = "/content/sam_vit_h_4b8939.pth"  # Update the path if needed
sam = sam_model_registry["vit_h"](checkpoint=sam_checkpoint)
sam_predictor = SamPredictor(sam)

def yolo_sam_hybrid(image_path):
    # Load image
    image = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Run YOLO detection
    results = yolo_model.predict(image_rgb, conf=0.5)

    # Extract YOLO detections (bounding boxes)
    for result in results:
        boxes = result.boxes.xyxy.cpu().numpy()  # Convert to NumPy array
        for box in boxes:
            x1, y1, x2, y2 = map(int, box)  # Convert coordinates

            # Set SAM input image
            sam_predictor.set_image(image_rgb)

            # Create mask with SAM using bounding box from YOLO
            input_box = np.array([x1, y1, x2, y2])
            masks, _, _ = sam_predictor.predict(box=input_box[None, :], multimask_output=True)

            # Draw YOLO bounding box
            cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Apply segmentation mask
            mask = masks[0]
            image[mask > 0] = [0, 0, 255]  # Red overlay for segmented area

    return image

#  Test on an image
result_img = yolo_sam_hybrid("/content/runs/detect/predict/frame_0100.jpg")

#  Save result
cv2.imwrite("yolo_sam_output.jpg", result_img)

#  Display in Colab
cv2_imshow(result_img)  # No window name needed in Colab
```

```python
import cv2
import os
import numpy as np
from ultralytics import YOLO
import torch

# Load YOLO trained model
yolo_model = YOLO("/content/content/runs/detect/train5/weights/best.torchscript")

# Load SAM model
from segment_anything import SamPredictor, sam_model_registry
sam_model = sam_model_registry["vit_h"](checkpoint="/content/sam_vit_h_4b8939.pth")  # Ensure correct checkpoint path
sam_model.to("cuda")  # Move SAM to GPU if available
sam_predictor = SamPredictor(sam_model)

# Path to input images
input_folder = "/content/drive/MyDrive/remaining_frames/"  # Update with your folder containing 1500 images
output_folder = "/content/drive/MyDrive/yolo_sam_output/"  # Folder to save segmented images
os.makedirs(output_folder, exist_ok=True)

# Function to run YOLO-SAM hybrid model
def yolo_sam_hybrid(image_path, save_path):
    # Load image
    image = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Run YOLO detection
    results = yolo_model.predict(image_rgb, conf=0.5)

    # Extract YOLO detections (bounding boxes)
    for result in results:
        boxes = result.boxes.xyxy.cpu().numpy()  # Convert to NumPy array

        for box in boxes:
            x1, y1, x2, y2 = map(int, box)  # Convert coordinates

            # Set SAM input image
            sam_predictor.set_image(image_rgb)

            # Create mask with SAM using bounding box from YOLO
            input_box = np.array([x1, y1, x2, y2])
            masks, _, _ = sam_predictor.predict(box=input_box[None, :], multimask_output=False)

            # Draw YOLO bounding box
            cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Apply segmentation mask
            mask = masks[0]
            image[mask > 0] = [0, 0, 255]  # Red overlay for segmented area

    # Save the processed image
    cv2.imwrite(save_path, image)

# Process all 1500 images
image_files = sorted(os.listdir(input_folder))[:1500]  # Get first 1500 images
for img_file in image_files:
    input_path = os.path.join(input_folder, img_file)
    output_path = os.path.join(output_folder, img_file)
    yolo_sam_hybrid(input_path, output_path)

print(" Saved 1500 YOLO-SAM processed images to:", output_folder)
```
