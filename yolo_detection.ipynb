```python
from ultralytics import YOLO

# Load YOLO model
model = YOLO("yolov8s.pt")

# Train YOLO with optimized hyperparameters
model.train(
    data="/content/drive/MyDrive/merged_dataset/data.yaml",  #  Dataset path
    epochs=100,  # More epochs for better convergence
    imgsz=640,  # Standard image size
    batch=16,  # Increase if GPU allows, otherwise reduce to 8
    patience=20,  # Stop early if no improvement
    optimizer="AdamW",  # Better for convergence
    lr0=0.001,  # Initial learning rate
    lrf=0.01,  # Final learning rate
    weight_decay=0.0005,  # Prevents overfitting
    momentum=0.937,  # Stabilizes training
    warmup_epochs=3,  # Helps in smooth training start
    warmup_momentum=0.8,  # Momentum during warmup
    dropout=0.1,  # Reduces overfitting
    device="cuda"  # Use GPU
)
```

```python
from ultralytics import YOLO

# Load best model from training
model = YOLO("runs/detect/train5/weights/best.pt")

# Export for future use
model.export(format="torchscript")  # Saves the model
```

```python
from ultralytics import YOLO

# Load the exported model
model = YOLO("runs/detect/train5/weights/best.torchscript")

# Run inference on test images
model.predict(source="/content/drive/MyDrive/merged_dataset/images/test", save=True, conf=0.5)
```

```python
model.val(data="/content/drive/MyDrive/merged_dataset/data.yaml", imgsz=640)
```
